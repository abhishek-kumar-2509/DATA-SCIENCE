```python
def add(*nums):
  total = 0
  for x in nums:
    total += x
  return (total)
result1 = add(10, 20)
print(result1)
result2 = add(1,20,30,40,50)
print(result2)
```

```python
import numpy as np
n = int(input("How many numbers: "))
numbers = []
for i in range(n):
  x = int(input("Enter a number: "))
  numbers.append(x)

arr = np.array(numbers)
print("Shape", arr.shape)
print("Dim", arr.ndim)
print("type", type(arr))
print("Array Type", arr.dtype)
```

```python
def create_Array(n):
  numbers = []
  for i in range(n):
    x = int(input("Enter a number: "))
    numbers.append(x)
  return np.array(numbers)
n = int(input("How many numbers: "))
arr = create_Array(n)
print("Shape", arr.shape)
print("Dim", arr.ndim)
print("type", type(arr))
print("Array Type", arr.dtype)
```

```python
# Inpute number of students
n = int(input("Enter number of students: "))

marks = []

# Input marks of each student
for i in range(n):
  m = float(input(f"Enter marks of students {i+1}: "))
  marks.append(m)

# Calculate minimum, maximum and average
minimum = min(marks)
maximum = max(marks)
Average = sum(marks)/n

# print result
print("Minimum marks: ", minimum)
print("Maximum marks: ", maximum)
print("Average marks: ", Average)
```

flatter() and ravel()

```python
import numpy as np
data = np.array([[1,2,3],[4,5,6]])
print("Orig data ", data.ndim)
flat1 = data.flatten()
print("Flatten flat1", flat1)
flat2 = data.ravel()
print("Ravel flat2", flat2)
```

Combining Arrays
# np.concatenate()
# np.vstack()
# np.hstack()


```python
a = np.array([1,2,3])
b = np.array([4,5,6])
c = np.concatenate((a,b))

print("Concatenate: \n", c)
print("vstack:  \n", np.vstack((a,b)))
print("hstack:  \n", np.hstack((a,b)))

```
