```python
# if.. ifelse.. ifelse(nested if else)
marks=75
if marks>=90:
  print("Grade A")
elif marks>=80:
  print("Grade B")
elif marks>=70:
  print("Grade B+")
elif marks>=60:
  print("Grade C")
elif marks>=45:
  print("Grade D")
else:
  print("Fail")
```

```python
# extend function
sec_a=["Chirag","Jhansi","Devi"]
sec_b=["Kiran","Ganesh","Lakshman"]
sec_a.extend(sec_b)
print(sec_a)
print(sec_b)
```

```python
# remove an item in the list
sec_a_student=["Chirag","Jhansi","Devi","KAlinga","Manish"]
sec_a_student.remove("Devi")
print(sec_a_student)
del sec_a_student[1]
print(sec_a_student)
sec_a_student.pop()
print(sec_a_student)
```
