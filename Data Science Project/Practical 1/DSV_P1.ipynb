```python
import pandas as pd
import numpy as np
```

```python
raw_students = pd. DataFrame({
"Roll_No": [1, 2, 3, 3, 4, 5],
"Name": [" aman", "PRIYA", "Rahul", "Rahul", "Neha", "Isha"],
"Department": ["cse", "IT", "CSE", "CSE", "ece", "IT"],
"Marks": ["85", "90", "78", "78", "missing", "67"],
"Attendance": [92, 95, 80, 80, 88, np.nan]
})
print(raw_students)
```

```python
print("Shape:",raw_students.shape)

print("\nData Types:")
print(raw_students.dtypes)

print("\nMissing Values:")
print(raw_students.isnull().sum())

print("\nDuplicate Records:")
print(raw_students.duplicated().sum())
```

```python
student = raw_students.copy()

student["Name"] = student["Name"].str.strip().str.title()

student["Department"] = student["Department"].str.strip().str.upper()

print(student)
```

```python
student["Marks"] = pd.to_numeric(student["Marks"], errors="coerce")

print(student)
```

```python
student = student.drop_duplicates(subset="Roll_No", keep="first")

print(student)
```

```python
student["Marks"] = student["Marks"].fillna(student["Marks"].median())

student["Attendance"] = student["Attendance"].fillna(student["Attendance"].median())

print(student)
```

```python
student["Pass"] = student["Marks"] >= 40

print(student)
```

```python
department_analysis = student.groupby("Department")[["Marks","Attendance"]].mean()

print(department_analysis)
```

```python
student.to_csv("cleaned_student.csv", index=False)

student.to_pickle("cleaned_student.pkl")
print("Files Save Successfully!")
```
