```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#  Visualization
# Graphs
# 1. Line graphs
# 2. Bar graph
# 3. Barh
# 4. Scatter
# 5. Box plot
# 6. Histogram
# 7. piechart
```

```python
# 1. Line Graph
subject = ['Python','Statistics',"DBMS",'Machin Learning','Data Visualization']
marks = [85,78,90,82,88]
```

```python
# Create the line graph
plt.plot(subject,marks)
#  Set the title
plt.title("Student Performance Analysis")

# Label the X-axis
plt.xlabel("Subject")

# Label the Y-axis
plt.ylabel("Marks")

# Apply the grid
plt.grid(True)

# Readability
plt.xticks(rotation=30)


# Display the graph
plt.show()
```

```python
plt.plot(subject,
         marks,
         color="green",
         linestyle="--",
         linewidth=5,
         marker="o", # marker type is circle |s|D|*
         label="Students",
         markersize=8,
         markerfacecolor="blue"
         )

plt.title("Student Performance Analysis")
plt.xlabel("subject")
plt.ylabel("Marks")
plt.xticks(rotation=30)
plt.grid(True)
plt.legend()
plt.show()
```

```python
# from matplotlib.lines import lineStyles
# Bar Chart
# Create
plt.figure(figsize=(8,5)) #width = 8, height = 5

plt.bar(
    subject,
    marks,
    color="pink",
    edgecolor="green",
    width=0.8
    )




plt.title("Bar Chart")
plt.xlabel("Subject")
plt.ylabel("Marks")
plt.xticks(rotation=20)
plt.grid(axis="y",alpha=0.5,linestyle="--")  # alpha it specifies the transparency

for i, value in enumerate(marks):
  plt.text(i, value+1, str(value), ha="center")

plt.ylim(0,100)

plt.show()
```

```python
# create the horixontal bar chart
plt.figure(figsize=(6,4))
plt.barh(
    subject,
    marks,
    color="orange"
)
plt.title("horizontal Bar chart for Students marks")
plt.show()
```

```python
# scatter Plot
# individual Observation
# Number of hours studied by the student
study_hours = [1, 2, 3, 4, 5, 6, 7, 8]
marks = [35, 40, 45, 50, 55, 68, 79, 90]

# create the Scatter plot
plt.figure(figsize=(8, 5))

plt.scatter(
    study_hours,
    marks,
    s=150,     # size
    c="red",    # red
    marker="o", #Type of marker is circle
    alpha=0.8,
    label="Student",
    zorder=3,   # Display the marker above the grid
)

plt.title("study_hours VS Examination Marks", fontsize=20)
plt.xlabel("Study Hours", fontsize=15, fontweight="bold")
plt.ylabel("Examination Marks", fontsize=15, fontweight="bold")

plt.xticks(study_hours)
plt.xlim(0, 9)
plt.ylim(0, 100)
plt.grid(True, linestyle="--", alpha=0.8)

plt.legend()
plt.show()
```
